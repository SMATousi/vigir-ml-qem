```python
import json, os
from tqdm import tqdm
```

```python
with open("../../../quantum/ExecutionResults/DataAggregation/data_aggregationJuly18_2025.json") as json_file:
    raw_data = json.load(json_file)
```

```python
os.makedirs("extracted_data", exist_ok=True)
```

```python
for data_dict in tqdm(raw_data):

    # print(data_dict.keys())
    json_file_name = f"{data_dict['pretranspiled_circuit']}_hardware_shots.json"
    extracted_data = {}

    extracted_data["counts"] = data_dict['job_measurements']

    file_path = os.path.join("extracted_data",json_file_name)
    
    with open(file_path, 'w') as json_file:
        json.dump(extracted_data, json_file)
    # print(extracted_data)
    # break
```
