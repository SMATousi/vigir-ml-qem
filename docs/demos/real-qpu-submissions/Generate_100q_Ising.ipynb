```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import random_clifford
from tqdm import tqdm

def generate_ising_twirl_circuits(num_qubits: int,
                                  num_circuits: int,
                                  depth: int = 10) -> list[QuantumCircuit]:
    """
    Generate a list of random 100-qubit Ising-twirl circuits.

    Parameters:
    - num_qubits: Number of qubits per circuit.
    - num_circuits: Total number of circuits to generate.
    - depth: Number of Trotter layers per circuit (default: 10).

    Returns:
    - circuits: List of QuantumCircuit objects.
    """
    circuits = []
    for idx in tqdm(range(num_circuits)):
        qc = QuantumCircuit(num_qubits)
        # Initial random single-qubit Clifford twirl
        for q in range(num_qubits):
            qc.append(random_clifford(1).to_instruction(), [q])
        # Ising Trotter layers with twirling
        for _layer in range(depth):
            # Random pairing (matching) of qubits for ZZ interactions
            perm = np.random.permutation(num_qubits)
            for i in range(0, num_qubits - 1, 2):
                q1, q2 = perm[i], perm[i + 1]
                qc.rzz(np.pi / 4, q1, q2)
            # Twirl between layers with single-qubit Cliffords
            for q in range(num_qubits):
                qc.append(random_clifford(1).to_instruction(), [q])
        circuits.append(qc)
    return circuits
```

```python
num_qubits = 100
num_circuits = 10
depth = 10  # adjust as needed
circuits = generate_ising_twirl_circuits(num_qubits, num_circuits, depth)
print(f"Generated {len(circuits)} Ising-twirl circuits with {num_qubits} qubits and depth {depth}")
```

```python
import numpy as np
import json
from qiskit import transpile
from qiskit_ibm_runtime import QiskitRuntimeService, Session
from qiskit_aer.primitives import Estimator as AerEstimator
from qiskit.quantum_info import SparsePauliOp
from mitiq import zne
from mitiq.zne.scaling.folding import fold_gates_at_random
from mitiq.zne import execute_with_zne
from mitiq.zne.scaling import fold_gates_at_random
from mitiq.zne.inference import RichardsonFactory
from qiskit_ibm_runtime import QiskitRuntimeService, Session, Estimator as RuntimeEstimator
from qiskit.quantum_info import SparsePauliOp
from mitiq.zne import execute_with_zne
from mitiq.zne.scaling import fold_gates_at_random
from mitiq.zne.inference import RichardsonFactory

def build_observables(num_qubits, measure_qubits):
    """Create Z Pauli observables on given qubits."""
    obs = []
    for q in measure_qubits:
        label = ['I'] * num_qubits
        label[q] = 'Z'
        obs.append(SparsePauliOp(''.join(label)))
    return obs

def run_zne_on_backend(circuits, backend_name, measure_qubits,
                       scale_factors, shots=8192):
    """
    Returns a dict mapping each circuit index to:
      {'noisy': [...], 'mitigated': [...]}
    """
    # 1. Get service and backend
    service = QiskitRuntimeService()                      # V2 primitives :contentReference[oaicite:7]{index=7}
    backend = service.backend(backend_name)
    circuits = transpile(circuits, backend=backend)

    # 2. Open a session
    with Session(backend) as session:
        estimator = RuntimeEstimator(session)             # pass Session as mode :contentReference[oaicite:8]{index=8}

        # 3. Set your shot count via options.default_shots
        estimator.options.default_shots = shots           # control shots here :contentReference[oaicite:9]{index=9}

        observables = build_observables(
            circuits[0].num_qubits, measure_qubits
        )
        results = {}

        # 4. Loop over circuits & observables
        for idx, circ in enumerate(circuits):
            noisy_vals = []
            mitigated_vals = []

            for obs in observables:
                # Raw (noisy) expectation
                job = estimator.run([(circ, obs)])           # uses default_shots :contentReference[oaicite:10]{index=10}
                raw = job.result().values[0]
                noisy_vals.append(raw)

                # ZNE-mitigated expectation
                fn = lambda c: estimator.run([c], [obs]) \
                                   .result().values[0]
                zne_val = execute_with_zne(
                    circuit=circ,
                    executor=fn,
                    scale_factors=scale_factors,
                    noise_scaling=fold_gates_at_random,
                    extrapolation_factory=RichardsonFactory()
                )
                mitigated_vals.append(zne_val)

            results[idx] = {
                'noisy': noisy_vals,
                'mitigated': mitigated_vals
            }

    return results


```
